# 00 - Environment check
Confirms required packages are installed before running the rest of the pipeline.

```python
import importlib.util

REQUIRED = ["numpy", "pandas", "scipy", "sklearn", "xgboost", "astropy", "matplotlib", "pyarrow"]

missing = [pkg for pkg in REQUIRED if importlib.util.find_spec(pkg) is None]
if missing:
    raise ImportError(f"Missing packages: {missing}. Run: pip install -r ../requirements.txt")
print("All required packages available.")
```
